# Errors in Rigid Registration

In the [previous section](#ref_Tracking) we presented {term}`OPA` as a means to achieve **rigid body registration** in the context of optical tracking, it has in fact a much wider applicability in all {term}`IGI` applications ([](https://doi.org/10.1007/978-0-387-73858-1_6)). For example, while the {term}`OPA` address the registration in a pair-wise fasion, it is a precursor for surface-based registration ([](https://doi.org/10.1109/34.121791)), both are either as an end in itself or as a precursor to nonrigid registration. Here, we will examine how to achieve the best rigid body registration in the context of minimizing errors.

## Fiducial Points

Let's solidify the concept of rigid registration with terminology. Using the context provided in [optical tracking using rigid body registration](#ref_Tracking), there are two types of data points:

1. *Model* point sets ($X_i$): geometry of points sets (i.e. the location of the [optical retroreflective spheres](https://www.ndigital.com/optical-navigation-technology/passive-marker-spheres/)) that are **assumed** to be error-free:
    - They were designed and manufactured with very high precision,
    - It is assumed that the manufactured device, i.e. the physical locations of these retroreflective spheres relative to each other, matches that with the {term}`CAD` model **exactly**, hence the name *model point* sets,
    - **In practice**, there will be deviation between the (manufactured) physical location with the {term}`CAD` model!
    - Other examples may includes:
        - Anatomical structures an expert clinician measured from (a **model**) medical imaging modalities such as {term}`CT` or {term}`MRI`.
        - Artificial markers attached to the patients. Example includes multi-modality makers such as [this](https://izimed.com/products/multi-modality-markers).
    - In the literature, they may be referred to as **reference points** or **reference fiducial points**. 
2. *Measurement* point sets ($Y_i$): geometry of these fiducial points measured (**localized**) in another modality such as
    - A tracking system,
    - Another image modality.

The purpose of rigid registration is to find a transformation (Rotation followed by Translation), that maps the <span style="color:blue">model fiducial point set</span> to those <span style="color:red">measurement point sets</span> in a different modality. In the process of doing so, we also want to find out the location of a target:

In this regard, we further differentiate between two types of fiducial points:
1. **Registration Fiducial** are fiducial points *whose locations are used to derive the rigid registration transformation*. Examples include
    - Anatomical structures that are easily accessible (e.g. tip of a nose),
    - Artificial markers attached to the patients, 
    - Retroreflective spheres of an optically tracked surgical stylus, and
1. **Target Fiducial** are fiducial points *whose location(s) is **not** used to derive the registration registration*, but are of some other interests. Examples include
    - Anatomical structures that are **not** easily accessible such as the location of a tumour,
    - The tip of an optically tracked surgical stylus.

## Fiducial Localization Error

In current literatures of {term}`IGI`,  the <span style="color:blue">model fiducial point set</span> is assumed to be error-free, i.e. their measurement **exact**, although in practice they are not.  On the other hand, when the registration fiducials are physically measured in another modality, such a via optical tracking, errors and noise in measurement will inevitably be introduced, resulting in **Fiducial Localization Error** ({term}`FLE`).

```{figure} ./../images/003/fig_rigid_registration_FLE.png
:label: fig_rigid_registration_FLE
:alt: Rigid registration
:align: center
:width: 80%

In this exggeratd example, the location of the <span style="color:red"> measured fiducial points set</span> contain errors: The solid spheres are in their **true** location, where as the dashed spheres are the measured (localized) locations. The Euclidean distance between the true and the measured location is termed Fiducial Localization Error ({term}`FLE`).
```

That is, the {term}`FLE` is the <wiki:euclidean_distance> btween the **forever unknown** ground truth value and the **known**, but with error, measurements.

## Fiducial Registration Error

## Target Localization Error and Target Registration Error
