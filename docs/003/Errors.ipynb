# Errors in Rigid Registration

In the [previous section](#ref_Tracking) we presented {term}`OPA` as a means to achieve **rigid body registration** in the context of optical tracking, it has in fact a much wider applicability in all {term}`IGI` applications ([](https://doi.org/10.1007/978-0-387-73858-1_6)). For example, while the {term}`OPA` address the registration in a pair-wise fasion, it is a precursor for surface-based registration ([](https://doi.org/10.1109/34.121791)), both are either as an end in itself or as a precursor to nonrigid registration. Here, we will examine how to achieve the best rigid body registration in the context of minimizing errors.

## Fiducial Points

Let's solidify the concept of rigid registration with terminology. Using the context provided in [optical tracking using rigid body registration](#ref_Tracking), there are two types of data points:

1. *Model* point sets ($X_i$): geometry of points sets (i.e. the location of the [optical retroreflective spheres](https://www.ndigital.com/optical-navigation-technology/passive-marker-spheres/)) that are **assumed** to be error-free:
    - They were designed and manufactured with very high precision,
    - It is assumed that the manufactured device, i.e. the physical locations of these retroreflective spheres relative to each other, matches that with the {term}`CAD` model **exactly**, hence the name *model point* sets,
    - **In practice**, there will be deviation between the (manufactured) physical location with the {term}`CAD` model!
    - Other examples may includes:
        - Anatomical structures an expert clinician measured from (a **model**) medical imaging modalities such as {term}`CT` or {term}`MRI`.
        - Artificial markers attached to the patients. Example includes multi-modality makers such as [this](https://izimed.com/products/multi-modality-markers).
    - In the literature, they may be referred to as **reference points** or **reference fiducial points**. 
2. *Measurement* point sets ($Y_i$): geometry of these fiducial points measured (**localized**) in another modality such as
    - A tracking system,
    - Another image modality.

In this regard, we further differentiate between two types of fiducial points:
1. **Registration Fiducial** are fiducial points *whose locations are used to derive the rigid registration transformation*. Examples include
    - Anatomical structures that are easily accessible (e.g. tip of a nose),
    - Artificial markers attached to the patients, 
    - Retroreflective spheres of an optically tracked surgical stylus, and
1. **Target Fiducial** are fiducial points *whose location(s) is **not** used to derive the registration registration*, but are of some other interests. Examples include
    - Anatomical structures that are **not** easily accessible such as the location of a tumour,
    - The tip of an optically tracked surgical stylus.

## Fiducial Localization Error

In current literatures of {term}`IGI`,  the <span style="color:blue">model fiducial point set</span> is assumed to be error-free, i.e. their measurement **exact**, although in practice they are not.  On the other hand, when the registration fiducials are physically measured in another modality, such a via optical tracking, errors and noise in measurement will inevitably be introduced, resulting in **Fiducial Localization Error** ({term}`FLE`).


That is, the {term}`FLE` is the <wiki:euclidean_distance> between the **forever unknown** ground truth value and the **known**, but with error, measurements.

Note that the ground-truth value can **never** be determined exactly as there *will always be error and noise* in the measurement. However, according to <wiki:Law_of_large_numbers>, if a sufficient large number of independent measurements are made, their **average* converges to the true value, if it exist. *Formally*, the <wiki:law_of_large_numbers> states that **given a sample of *independent* and *identically distributed* values, the sample mean converges to the 

(ref_Registration_Errors)=
## Fiducial Registration Error and Target Registration Error

Given two **homologous** point sets, the (perfect) model points $X_i$ and the (contaminated) measured points $Y_i$, the objective of the rigid-body transform is to find a rotation and translation that minimizes the <wiki:Euclidean_distance> between the model point set and the measurement point set after the registration:

$$
{FRE}^2 = \frac{1}{N} \sum_{i=1}^N \left| R x_i + t - y_i \right|^2
$$
where $N$ is the <wiki:cardinality> of the point sets (and **homologous** implies that there is a *one-to-one correspondence*, implying that both point sets have the **same** cardinality), $R$ and $t$ are the rotation and translation to be determined, respectively, and {term}`FRE` (or Fiducial Registration Error) is the averaged <wiki:Euclidean_distance> between each paired-points after the registration.

The closed-form {term}`OPA` solution introduced [earlier](#ref_OPA) is, in fact, the **optimal** solution in the [L2-norm](https://en.wikipedia.org/wiki/Norm_(mathematics)#p-norm) sense, provided the errors in the measured points are [zero-mean, isotropic, independent, and identically distributed](https://en.wikipedia.org/wiki/Independent_and_identically_distributed_random_variables).

```{figure} ./../images/003/fig_rigid_registration_TRE.png
:label: fig_rigid_registration TRE
:alt: Relationship between TRE and FRE
:align: center
:width: 80%

Rigid Registration transforms <span style="color:blue"> one point set</span>, using rotation and translation, to <span style="color:red"> another point set</span>. The <wiki:Euclidean_distance> between the registration fiducial points is termed Fiducial Registration Error ({terma}`FRE`), where as the <wiki:Euclidean_distance> between the target fiducial points is termed Target Registration Error ({term}`TRE`). Ths misalignments after registration are the results of **Fiducial Localization Error** ({term}`FLE`).
```

Concretely, let $x_i$ be the $i^{th}$ registration fiducial and $p$ be the target point. Then,

$$
{FRE}_i = \left| R x_i + t - y_i \right|
$$
is the <wiki:Euclidean_distance> between $x_i$ and $y_i$ after registration, and

$$
TRE(q) = \left| R p + t - q \right|
$$
where $q$ is the transformed point $p$, and TRE(q) is the <wiki:Euclidean_disance> of $p$ and $q$ after the transformation, and

$$
{FRE}^2 = \frac{1}{N} \sum_{i=1}^N FRE_i^2
$$

Again, it should be noted that the [closed-form solution](#ref_OPA) is optimal assuming that the original errors are isotropic, independent and identically distributed ({term}`IID`) with a zero mean [Gaussian distribution](https://en.wikipedia.org/wiki/Normal_distribution). In most cases, however, these assumptions do not hold. 

For example, measurements from optical tracking system are *anisotropic* with the uncertainly larger in the axis parallel to the viewing angle of the camera, but nonetheless the [closed-form solution](#ref_OPA) serves as a reasonable approximation to the true solution.
